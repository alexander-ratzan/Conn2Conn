## Latent Inspection
Focus the second half of the notebook on PCA score structure, token construction, and attention before vs after training.

```python

import torch
import numpy as np
import matplotlib.pyplot as plt

from models.config import build_model
from models.loss import get_model_input

model_trained = run_out["model"]
model_trained.eval()

torch.manual_seed(0)
model_init = build_model(sim.base, "LatentAttnMasked", config_override["model"])
model_init.eval()

val_batch = next(iter(sim.val_loader))
x_val = get_model_input(val_batch)
y_val = val_batch["y"]

n_mean_subjects = min(16, x_val.shape[0])

x_example = x_val[:1]
y_example = y_val[:1]
x_mean_batch = x_val[:n_mean_subjects]
y_mean_batch = y_val[:n_mean_subjects]

with torch.no_grad():
    inspect_init_ex = model_init.inspect_attention_state(x_example, y=y_example, force_all_masked=True)
    inspect_trained_ex = model_trained.inspect_attention_state(x_example, y=y_example, force_all_masked=True)
    inspect_init_mean = model_init.inspect_attention_state(x_mean_batch, y=y_mean_batch, force_all_masked=True)
    inspect_trained_mean = model_trained.inspect_attention_state(x_mean_batch, y=y_mean_batch, force_all_masked=True)

latent_debug = {
    "sc_pca_scores": model_trained.source_pca_scores_train.detach().cpu(),
    "fc_pca_scores": model_trained.target_pca_scores_train.detach().cpu(),
    "sc_pca_scores_z": model_trained.source_pca_scores_train_z.detach().cpu(),
    "fc_pca_scores_z": model_trained.target_pca_scores_train_z.detach().cpu(),
    "inspect_init_ex": {k: (v.detach().cpu() if torch.is_tensor(v) else v) for k, v in inspect_init_ex.items()},
    "inspect_trained_ex": {k: (v.detach().cpu() if torch.is_tensor(v) else v) for k, v in inspect_trained_ex.items()},
    "inspect_init_mean": {k: (v.detach().cpu() if torch.is_tensor(v) else v) for k, v in inspect_init_mean.items()},
    "inspect_trained_mean": {k: (v.detach().cpu() if torch.is_tensor(v) else v) for k, v in inspect_trained_mean.items()},
}

{k: (tuple(v.shape) if hasattr(v, "shape") else sorted(v.keys())) for k, v in latent_debug.items()}
```

```python

# PCA score structure: subject-by-component scores and joint component covariance
sc_scores = latent_debug["sc_pca_scores"].numpy()
fc_scores = latent_debug["fc_pca_scores"].numpy()

joint_scores = np.concatenate([sc_scores, fc_scores], axis=1)
joint_cov = np.cov(joint_scores, rowvar=False)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

im0 = axes[0, 0].imshow(sc_scores, aspect="auto", cmap="RdBu_r")
axes[0, 0].set_title("SC PCA Scores")
axes[0, 0].set_xlabel("Component")
axes[0, 0].set_ylabel("Subject")
plt.colorbar(im0, ax=axes[0, 0], fraction=0.03, pad=0.03)

im1 = axes[0, 1].imshow(fc_scores, aspect="auto", cmap="RdBu_r")
axes[0, 1].set_title("FC PCA Scores")
axes[0, 1].set_xlabel("Component")
axes[0, 1].set_ylabel("Subject")
plt.colorbar(im1, ax=axes[0, 1], fraction=0.03, pad=0.03)

axes[1, 0].bar(np.arange(sc_scores.shape[1]), sc_scores.var(axis=0), color="#3b82f6", alpha=0.8)
axes[1, 0].set_title("SC PCA Variance by Component")
axes[1, 0].set_xlabel("Component")
axes[1, 0].set_ylabel("Variance")

im3 = axes[1, 1].imshow(joint_cov, cmap="coolwarm")
axes[1, 1].set_title("Joint SC/FC PCA Covariance")
axes[1, 1].set_xlabel("Component (SC then FC)")
axes[1, 1].set_ylabel("Component (SC then FC)")
axes[1, 1].axvline(sc_scores.shape[1] - 0.5, color="k", linestyle="--", linewidth=1)
axes[1, 1].axhline(sc_scores.shape[1] - 0.5, color="k", linestyle="--", linewidth=1)
plt.colorbar(im3, ax=axes[1, 1], fraction=0.03, pad=0.03)

plt.tight_layout()
plt.show()
```

```python

# Initial vs trained token states for one example validation subject
example_init = latent_debug["inspect_init_ex"]
example_trained = latent_debug["inspect_trained_ex"]

def show_token_panel(title, tensor, vmax=None):
    arr = tensor.squeeze(0).numpy()
    if vmax is None:
        vmax = np.max(np.abs(arr)) + 1e-8
    plt.imshow(arr, aspect="auto", cmap="RdBu_r", vmin=-vmax, vmax=vmax)
    plt.title(title)
    plt.xlabel("Token feature")
    plt.ylabel("Token index (SC then FC)")
    plt.axhline(model_trained.n_components_pca - 0.5, color="k", linestyle="--", linewidth=1)
    plt.colorbar(fraction=0.03, pad=0.03)

fig = plt.figure(figsize=(14, 10))
plt.subplot(2, 2, 1)
show_token_panel("Initial Tokens (Pre-Attention)", example_init["tokens_pre"])
plt.subplot(2, 2, 2)
show_token_panel("Initial Tokens (Post-Attention)", example_init["tokens_post"])
plt.subplot(2, 2, 3)
show_token_panel("Trained Tokens (Pre-Attention)", example_trained["tokens_pre"])
plt.subplot(2, 2, 4)
show_token_panel("Trained Tokens (Post-Attention)", example_trained["tokens_post"])
plt.tight_layout()
plt.show()
```

```python

# Attention before vs after training: example subject and fast mean summary
example_init_attn = latent_debug["inspect_init_ex"]["attention"].squeeze(0).numpy()
example_trained_attn = latent_debug["inspect_trained_ex"]["attention"].squeeze(0).numpy()
mean_init_attn = latent_debug["inspect_init_mean"]["attention"].mean(dim=0).numpy()
mean_trained_attn = latent_debug["inspect_trained_mean"]["attention"].mean(dim=0).numpy()

vmax = max(
    np.max(np.abs(example_init_attn)),
    np.max(np.abs(example_trained_attn)),
    np.max(np.abs(mean_init_attn)),
    np.max(np.abs(mean_trained_attn)),
) + 1e-8
vmin = 0.0 if config_override["model"].get("attention_activation", "softmax") == "softmax" else -vmax
vmax_use = 1.0 if config_override["model"].get("attention_activation", "softmax") == "softmax" else vmax

fig, axes = plt.subplots(2, 2, figsize=(14, 12))
for ax, mat, title in [
    (axes[0, 0], example_init_attn, "Initial Attention (Example Subject)"),
    (axes[0, 1], example_trained_attn, "Trained Attention (Example Subject)"),
    (axes[1, 0], mean_init_attn, f"Initial Attention Mean (First {n_mean_subjects} Val Subjects)"),
    (axes[1, 1], mean_trained_attn, f"Trained Attention Mean (First {n_mean_subjects} Val Subjects)"),
]:
    im = ax.imshow(mat, cmap="viridis", vmin=vmin, vmax=vmax_use)
    ax.set_title(title)
    ax.set_xlabel("Key token")
    ax.set_ylabel("Query token")
    ax.axvline(model_trained.n_components_pca - 0.5, color="w", linestyle="--", linewidth=1)
    ax.axhline(model_trained.n_components_pca - 0.5, color="w", linestyle="--", linewidth=1)
    plt.colorbar(im, ax=ax, fraction=0.03, pad=0.03)
plt.tight_layout()
plt.show()
```

```python
# Attention before vs after training with data-driven scaling
example_init_attn = latent_debug["inspect_init_ex"]["attention"]
example_trained_attn = latent_debug["inspect_trained_ex"]["attention"]
mean_init_attn = latent_debug["inspect_init_mean"]["attention"]
mean_trained_attn = latent_debug["inspect_trained_mean"]["attention"]

if example_init_attn is None or example_trained_attn is None:
    print("Attention matrix is unavailable for this run.")
else:
    example_init_attn = example_init_attn.squeeze(0).numpy()
    example_trained_attn = example_trained_attn.squeeze(0).numpy()
    mean_init_attn = mean_init_attn.mean(dim=0).numpy()
    mean_trained_attn = mean_trained_attn.mean(dim=0).numpy()

    mats = [
        example_init_attn,
        example_trained_attn,
        mean_init_attn,
        mean_trained_attn,
    ]

    if config_override["model"].get("attention_activation", "softmax") == "softmax":
        all_vals = np.concatenate([m.ravel() for m in mats])
        vmin = 0.0
        vmax_use = np.quantile(all_vals, 0.995)  # much better than forcing vmax=1
        cmap = "viridis"
    else:
        all_vals = np.concatenate([m.ravel() for m in mats])
        bound = np.quantile(np.abs(all_vals), 0.995)
        vmin = -bound
        vmax_use = bound
        cmap = "RdBu_r"

    print("Attention stats:")
    for name, mat in [
        ("example_init", example_init_attn),
        ("example_trained", example_trained_attn),
        ("mean_init", mean_init_attn),
        ("mean_trained", mean_trained_attn),
    ]:
        print(
            name,
            "min=", float(mat.min()),
            "max=", float(mat.max()),
            "mean=", float(mat.mean()),
        )

    fig, axes = plt.subplots(2, 2, figsize=(14, 12))
    for ax, mat, title in [
        (axes[0, 0], example_init_attn, "Initial Attention (Example Subject)"),
        (axes[0, 1], example_trained_attn, "Trained Attention (Example Subject)"),
        (axes[1, 0], mean_init_attn, f"Initial Attention Mean (First {n_mean_subjects} Val Subjects)"),
        (axes[1, 1], mean_trained_attn, f"Trained Attention Mean (First {n_mean_subjects} Val Subjects)"),
    ]:
        im = ax.imshow(mat, cmap=cmap, vmin=vmin, vmax=vmax_use, aspect="auto")
        ax.set_title(title)
        ax.set_xlabel("Key token")
        ax.set_ylabel("Query token")
        ax.axvline(model_trained.n_components_pca - 0.5, color="w", linestyle="--", linewidth=1)
        ax.axhline(model_trained.n_components_pca - 0.5, color="w", linestyle="--", linewidth=1)
        plt.colorbar(im, ax=ax, fraction=0.03, pad=0.03)

    plt.tight_layout()
    plt.show()
```
