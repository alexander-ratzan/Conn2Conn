## What we considered and rejected

| Idea | Why rejected |
|---|---|
| Task-aware objectives (predict cognition directly from FC/SC, end-to-end) | Demographic + brain-volume baseline already captures most of the cognition signal that survives residualization; method changes hit an information ceiling that is structural, not algorithmic |
| Fancier regressors (kernel methods, gradient boosting, deep) | The PCA+PLS / BR-per-component pipeline already saturates the variance explainable from FC/SC at the closed-form level; marginal gains from method changes have not materially moved cognition R² in any of the four prior attempts |
| BV-targeted optimization | BV is already a strong cognition predictor; the question is what *adds* on top, not what BV does alone — and the answer (FC for crystallized, ~nothing for SC) is already nailed down |
| Direct cognition-prediction benchmarks vs Krakencoder | Comparable to within noise; not citable as a separate contribution |

## Findings already in hand that close this thread

From the main notebook + tier_extensions:

- **FC**: substitution fidelity for crystallized cognition = 0.73× after demographic
  residualization. Source-modality (real FC → cog) and synthetic-from-SC (pred FC → cog)
  show the same ranking; FC carries real, non-demographic crystallized-cognition signal.
- **SC**: substitution fidelity ≈ 0× after residualization. SC's cognition signal IS
  the demographic signal — nothing left after controls.
- **BV**: a structural T1 alone (16 FreeSurfer volumes + age + sex + race/eth)
  achieves ~70% of the FC+BV+demo combined cognition R². Most of the cognition signal
  is structural anatomy.

These three together close the cognition-prediction story for this project: FC is
the modality that adds *non-anatomical* cognition signal; SC adds essentially nothing
beyond anatomy; chasing more cognition prediction with method changes is a no-upside drill.
